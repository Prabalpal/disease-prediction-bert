```python
import numpy as np
import torch
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

from datasets import Dataset, load_dataset
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    DataCollatorWithPadding,
    TrainingArguments,
    Trainer
)

# Load dataset (cached if already downloaded)
ds = load_dataset("gretelai/symptom_to_diagnosis")

train_df = ds["train"].to_pandas()
test_df = ds["test"].to_pandas()

train_df = train_df.rename(columns={
    "input_text": "text",
    "output_text": "disease"
})

test_df = test_df.rename(columns={
    "input_text": "text",
    "output_text": "disease"
})

# Create disease label IDs
labels = sorted(train_df["disease"].unique())
label2id = {name: i for i, name in enumerate(labels)}
id2label = {i: name for name, i in label2id.items()}

train_df["labels"] = train_df["disease"].map(label2id)
test_df["labels"] = test_df["disease"].map(label2id)

assert test_df["labels"].notna().all()

# Keep the original test set untouched
train_part, val_part = train_test_split(
    train_df[["text", "labels"]],
    test_size=0.15,
    stratify=train_df["labels"],
    random_state=42
)

test_part = test_df[["text", "labels"]].copy()

print("Training examples:", len(train_part))
print("Validation examples:", len(val_part))
print("Test examples:", len(test_part))
print("Disease classes:", len(labels))
print("GPU:", torch.cuda.get_device_name(0))

MODEL_NAME = "google-bert/bert-base-uncased"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=len(labels),
    id2label=id2label,
    label2id=label2id
)

def tokenize(batch):
    return tokenizer(
        batch["text"],
        truncation=True,
        max_length=128
    )

def prepare_dataset(frame):
    data = Dataset.from_pandas(
        frame.reset_index(drop=True),
        preserve_index=False
    )
    return data.map(
        tokenize,
        batched=True,
        remove_columns=["text"]
    )

train_ds = prepare_dataset(train_part)
val_ds = prepare_dataset(val_part)
test_ds = prepare_dataset(test_part)

collator = DataCollatorWithPadding(tokenizer=tokenizer)

print("Data preparation complete!")
```

```python
def compute_metrics(pred):
    logits = pred.predictions
    if isinstance(logits, tuple):
        logits = logits[0]

    predictions = np.argmax(logits, axis=-1)
    targets = pred.label_ids

    return {
        "accuracy": accuracy_score(targets, predictions),
        "precision_macro": precision_score(
            targets, predictions, average="macro",
            zero_division=0
        ),
        "recall_macro": recall_score(
            targets, predictions, average="macro",
            zero_division=0
        ),
        "f1_macro": f1_score(
            targets, predictions, average="macro",
            zero_division=0
        )
    }

training_args = TrainingArguments(
    output_dir="/content/disease_bert_checkpoints",
    learning_rate=2e-5,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=16,
    num_train_epochs=3,
    weight_decay=0.01,
    eval_strategy="epoch",
    save_strategy="epoch",
    load_best_model_at_end=True,
    metric_for_best_model="f1_macro",
    greater_is_better=True,
    save_total_limit=1,
    fp16=True,
    logging_steps=10,
    report_to="none",
    seed=42
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_ds,
    eval_dataset=val_ds,
    processing_class=tokenizer,
    data_collator=collator,
    compute_metrics=compute_metrics
)

trainer.train()
```

```python
# Evaluate on the original held-out test set
results = trainer.evaluate(
    eval_dataset=test_ds,
    metric_key_prefix="test"
)

print("\nFinal test metrics:")
for key, value in results.items():
    if isinstance(value, (int, float)):
        print(f"{key}: {value:.4f}")

pred = trainer.predict(test_ds)
predictions = np.argmax(pred.predictions, axis=-1)

print("\nPer-disease results:")
print(classification_report(
    pred.label_ids,
    predictions,
    labels=np.arange(len(labels)),
    target_names=labels,
    zero_division=0
))

# Save model and tokenizer
SAVE_PATH = "/content/disease_bert_model"

trainer.save_model(SAVE_PATH)
tokenizer.save_pretrained(SAVE_PATH)

print("\nModel saved at:", SAVE_PATH)
```

```python
import torch
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification
)

MODEL_PATH = "/content/disease_bert_model"

tokenizer = AutoTokenizer.from_pretrained(MODEL_PATH)
model = AutoModelForSequenceClassification.from_pretrained(MODEL_PATH)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)
model.eval()

def predict_disease(symptoms, top_k=3):
    inputs = tokenizer(
        symptoms,
        return_tensors="pt",
        truncation=True,
        max_length=128
    )
    inputs = {k: v.to(device) for k, v in inputs.items()}

    with torch.no_grad():
        logits = model(**inputs).logits[0]
        probabilities = torch.softmax(logits, dim=-1)

    scores, indices = torch.topk(
        probabilities,
        k=min(top_k, probabilities.numel())
    )

    results = []
    for score, idx in zip(scores.tolist(), indices.tolist()):
        disease = model.config.id2label.get(
            idx, model.config.id2label.get(str(idx), str(idx))
        )
        results.append({
            "predicted_label": disease,
            "model_score": round(score, 4)
        })

    return results

symptoms = (
    "I have a headache, fever, fatigue and a persistent cough."
)

for result in predict_disease(symptoms):
    print(result)
```

```python
from pathlib import Path
import shutil

from safetensors.torch import load_file, save_file
from transformers import AutoModelForSequenceClassification

MODEL_DIR = Path("/content/disease_bert_model_trial2")
WEIGHTS_FILE = MODEL_DIR / "model.safetensors"
BACKUP_FILE = MODEL_DIR / "model_before_keyfix.safetensors"

# 1. Back up the original checkpoint
if not BACKUP_FILE.exists():
    shutil.copy2(WEIGHTS_FILE, BACKUP_FILE)

# 2. Rename legacy LayerNorm parameters
state = load_file(str(WEIGHTS_FILE))
fixed_state = {}
renamed = 0

for key, tensor in state.items():
    new_key = key

    if key.endswith(".gamma"):
        new_key = key[:-len(".gamma")] + ".weight"
    elif key.endswith(".beta"):
        new_key = key[:-len(".beta")] + ".bias"

    if new_key in fixed_state:
        raise ValueError(f"Duplicate key after conversion: {new_key}")

    fixed_state[new_key] = tensor

    if new_key != key:
        renamed += 1

# Save the corrected weights
save_file(fixed_state, str(WEIGHTS_FILE))
print("Parameter names corrected:", renamed)
print("Original weights backed up:", BACKUP_FILE)

# 3. Reload the model and inspect compatibility
reloaded_model, info = (
    AutoModelForSequenceClassification.from_pretrained(
        str(MODEL_DIR),
        output_loading_info=True
    )
)

print("\nMissing keys:", info["missing_keys"])
print("Unexpected keys:", info["unexpected_keys"])
print("Error messages:", info.get("error_msgs", []))
```

```python
import torch
import gradio as gr
from transformers import AutoTokenizer, AutoModelForSequenceClassification

# Load the saved model
MODEL_PATH = "/content/drive/MyDrive/disease_bert_model_final"

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

tokenizer = AutoTokenizer.from_pretrained(MODEL_PATH)
model = AutoModelForSequenceClassification.from_pretrained(MODEL_PATH)
model.to(device)
model.eval()

print("Model loaded successfully!")
print("Device:", device)


def predict_disease(symptoms):
    if not symptoms or not symptoms.strip():
        return [["Please enter a symptom description.", None]]

    inputs = tokenizer(
        symptoms.strip(),
        return_tensors="pt",
        truncation=True,
        max_length=128
    )
    inputs = {key: value.to(device) for key, value in inputs.items()}

    with torch.no_grad():
        outputs = model(**inputs)
        scores = torch.softmax(outputs.logits[0], dim=-1)

    top_scores, top_indices = torch.topk(
        scores,
        k=min(3, scores.numel())
    )

    results = []

    for score, idx in zip(top_scores.tolist(), top_indices.tolist()):
        label = model.config.id2label.get(
            idx,
            model.config.id2label.get(str(idx), f"Class {idx}")
        )
        results.append([label, round(score, 4)])

    return results


# Create the interface
with gr.Blocks(title="BERT Disease Prediction Demo") as demo:
    gr.Markdown("# Disease Prediction Using Fine-Tuned BERT")
    gr.Markdown(
        "Enter a symptom description to view the model's top three "
        "candidate labels."
    )

    symptoms_input = gr.Textbox(
        label="Describe your symptoms",
        placeholder="Example: headache, fever, fatigue and cough",
        lines=4
    )

    predict_button = gr.Button("Predict", variant="primary")

    predictions_output = gr.Dataframe(
        headers=["Candidate label", "Model score (not a clinical probability)"],
        datatype=["str", "number"],
        label="Model predictions",
        interactive=False
    )

    predict_button.click(
        fn=predict_disease,
        inputs=symptoms_input,
        outputs=predictions_output
    )

    gr.Markdown(
        "**Important:** This is an educational research demonstration, "
        "not a medically validated diagnostic tool. Do not use its outputs "
        "to diagnose or treat a health condition."
    )

demo.launch(share=True)
```

```python
import os
import zipfile
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)
from google.colab import files

# ==========================================================
# 1. OUTPUT DIRECTORY
# ==========================================================
OUT = Path("/content/disease_prediction_submission")
OUT.mkdir(parents=True, exist_ok=True)

# ==========================================================
# 2. VALIDATION HISTORY FROM YOUR ACTUAL EXPERIMENT 2 LOG
# ==========================================================
history = pd.DataFrame({
    "epoch": [1, 2, 3, 4, 5],
    "validation_loss": [
        2.900467, 2.453568, 2.175125, 1.992546, 1.921318
    ],
    "accuracy": [
        0.140625, 0.546875, 0.664062, 0.757812, 0.773438
    ],
    "precision_macro": [
        0.129590, 0.582758, 0.720887, 0.774415, 0.782280
    ],
    "recall_macro": [
        0.137879, 0.545455, 0.663636, 0.760606, 0.775758
    ],
    "f1_macro": [
        0.080166, 0.494948, 0.631950, 0.733580, 0.746815
    ]
})

history.to_csv(OUT / "validation_history.csv", index=False)

# Graph 1: Validation loss
plt.figure(figsize=(8, 5))
plt.plot(
    history["epoch"],
    history["validation_loss"],
    marker="o"
)
plt.xlabel("Epoch")
plt.ylabel("Validation loss")
plt.title("Experiment 2: Validation Loss Across Epochs")
plt.xticks(history["epoch"])
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig(OUT / "01_validation_loss.png", dpi=300)
plt.show()

# Graph 2: Validation metrics across epochs
plt.figure(figsize=(9, 5))
for metric in [
    "accuracy", "precision_macro", "recall_macro", "f1_macro"
]:
    plt.plot(
        history["epoch"],
        history[metric],
        marker="o",
        label=metric.replace("_", " ").title()
    )

plt.xlabel("Epoch")
plt.ylabel("Score")
plt.title("Experiment 2: Validation Performance Across Epochs")
plt.xticks(history["epoch"])
plt.ylim(0, 1)
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig(OUT / "02_validation_metrics.png", dpi=300)
plt.show()

# ==========================================================
# 3. EVALUATE THE CORRECTED SAVED EXPERIMENT 2 MODEL
# ==========================================================
pred_output = verify_trainer.predict(test_ds)

logits = pred_output.predictions
if isinstance(logits, tuple):
    logits = logits[0]

y_true = pred_output.label_ids
y_pred = np.argmax(logits, axis=1)

num_classes = len(labels)
class_ids = np.arange(num_classes)

test_accuracy = accuracy_score(y_true, y_pred)
test_precision = precision_score(
    y_true, y_pred, average="macro", zero_division=0
)
test_recall = recall_score(
    y_true, y_pred, average="macro", zero_division=0
)
test_f1 = f1_score(
    y_true, y_pred, average="macro", zero_division=0
)

# Top-3 accuracy
top3_indices = np.argsort(logits, axis=1)[:, -3:]
top3_accuracy = np.mean([
    actual in candidates
    for actual, candidates in zip(y_true, top3_indices)
])

test_loss = pred_output.metrics.get("test_loss", np.nan)

# ==========================================================
# 4. METRICS COMPARISON: BOTH EXPERIMENTS
# ==========================================================
comparison = pd.DataFrame([
    {
        "Experiment": "Experiment 1",
        "Test loss": 1.928015,
        "Accuracy": 0.735849,
        "Macro precision": 0.804976,
        "Macro recall": 0.735678,
        "Macro F1": 0.720899
    },
    {
        "Experiment": "Experiment 2",
        "Test loss": test_loss,
        "Accuracy": test_accuracy,
        "Macro precision": test_precision,
        "Macro recall": test_recall,
        "Macro F1": test_f1
    }
])

comparison.to_csv(OUT / "experiment_comparison.csv", index=False)

print("TEST RESULTS — EXPERIMENT 2")
print(f"Test loss:          {test_loss:.4f}")
print(f"Accuracy:           {test_accuracy:.4f}")
print(f"Macro precision:    {test_precision:.4f}")
print(f"Macro recall:       {test_recall:.4f}")
print(f"Macro F1:            {test_f1:.4f}")
print(f"Top-3 accuracy:     {top3_accuracy:.4f}")

# Graph 3: Test-metric comparison
metric_cols = [
    "Accuracy", "Macro precision", "Macro recall", "Macro F1"
]

ax = comparison.set_index("Experiment")[metric_cols].plot(
    kind="bar",
    figsize=(10, 6)
)
ax.set_title("Test Performance: Experiment 1 vs Experiment 2")
ax.set_ylabel("Score")
ax.set_ylim(0, 1)
ax.set_xlabel("")
ax.legend(loc="lower right")
ax.grid(axis="y", alpha=0.3)
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig(OUT / "03_experiment_comparison.png", dpi=300)
plt.show()

# ==========================================================
# 5. CLASSIFICATION REPORT AND PER-DISEASE METRICS
# ==========================================================
report = classification_report(
    y_true,
    y_pred,
    labels=class_ids,
    target_names=labels,
    output_dict=True,
    zero_division=0
)

per_class = pd.DataFrame([
    {
        "Disease": disease,
        "Precision": report[disease]["precision"],
        "Recall": report[disease]["recall"],
        "F1 score": report[disease]["f1-score"],
        "Support": int(report[disease]["support"])
    }
    for disease in labels
])

per_class = per_class.sort_values("F1 score", ascending=False)
per_class.to_csv(OUT / "per_disease_scores.csv", index=False)

with open(OUT / "classification_report.txt", "w") as f:
    f.write(classification_report(
        y_true,
        y_pred,
        labels=class_ids,
        target_names=labels,
        zero_division=0
    ))

print("\nPER-DISEASE CLASSIFICATION REPORT")
print(per_class.to_string(index=False))

# Graph 4: F1-score per disease
plot_data = per_class.sort_values("F1 score", ascending=True)

plt.figure(figsize=(10, 9))
plt.barh(plot_data["Disease"], plot_data["F1 score"])
plt.xlabel("F1 score")
plt.ylabel("Disease class")
plt.title("Experiment 2: F1 Score by Disease")
plt.xlim(0, 1)
plt.grid(axis="x", alpha=0.3)
plt.tight_layout()
plt.savefig(OUT / "04_per_disease_f1.png", dpi=300)
plt.show()

# Graph 5: Per-class precision and recall
plot_data = per_class.sort_values("F1 score", ascending=True)
y_pos = np.arange(len(plot_data))
height = 0.38

plt.figure(figsize=(10, 10))
plt.barh(
    y_pos - height / 2,
    plot_data["Precision"],
    height,
    label="Precision"
)
plt.barh(
    y_pos + height / 2,
    plot_data["Recall"],
    height,
    label="Recall"
)
plt.yticks(y_pos, plot_data["Disease"])
plt.xlabel("Score")
plt.title("Experiment 2: Precision and Recall by Disease")
plt.xlim(0, 1)
plt.legend()
plt.grid(axis="x", alpha=0.3)
plt.tight_layout()
plt.savefig(OUT / "05_precision_recall_by_disease.png", dpi=300)
plt.show()

# ==========================================================
# 6. CONFUSION MATRIX
# ==========================================================
cm = confusion_matrix(y_true, y_pred, labels=class_ids)

pd.DataFrame(
    cm, index=labels, columns=labels
).to_csv(OUT / "confusion_matrix_counts.csv")

row_sums = cm.sum(axis=1, keepdims=True)
cm_normalized = cm / np.maximum(row_sums, 1)

plt.figure(figsize=(15, 13))
plt.imshow(cm_normalized, interpolation="nearest", vmin=0, vmax=1)
plt.colorbar(label="Proportion of actual class")
plt.xticks(class_ids, labels, rotation=90, fontsize=8)
plt.yticks(class_ids, labels, fontsize=8)
plt.xlabel("Predicted disease")
plt.ylabel("Actual disease")
plt.title("Experiment 2: Normalized Confusion Matrix")
plt.tight_layout()
plt.savefig(OUT / "06_normalized_confusion_matrix.png", dpi=300)
plt.show()

# ==========================================================
# 7. MOST COMMON MISCLASSIFICATIONS
# ==========================================================
errors = []

for i, actual in enumerate(labels):
    for j, predicted in enumerate(labels):
        if i != j and cm[i, j] > 0:
            errors.append({
                "Actual disease": actual,
                "Predicted disease": predicted,
                "Count": int(cm[i, j])
            })

errors_df = pd.DataFrame(errors)

if not errors_df.empty:
    errors_df = errors_df.sort_values("Count", ascending=False)
else:
    errors_df = pd.DataFrame(
        columns=["Actual disease", "Predicted disease", "Count"]
    )

errors_df.to_csv(OUT / "common_misclassifications.csv", index=False)

top_errors = errors_df.head(10).copy()

if not top_errors.empty:
    top_errors["Pair"] = (
        top_errors["Actual disease"]
        + " → "
        + top_errors["Predicted disease"]
    )
    top_errors = top_errors.sort_values("Count", ascending=True)

    plt.figure(figsize=(10, 6))
    plt.barh(top_errors["Pair"], top_errors["Count"])
    plt.xlabel("Number of test examples")
    plt.title("Most Common Misclassifications")
    plt.tight_layout()
    plt.savefig(OUT / "07_common_misclassifications.png", dpi=300)
    plt.show()

# ==========================================================
# 8. SAVE AN OVERALL SCORE SUMMARY
# ==========================================================
summary = pd.DataFrame([
    {"Metric": "Dataset training examples", "Value": 853},
    {"Metric": "Training examples used after split", "Value": 725},
    {"Metric": "Validation examples", "Value": 128},
    {"Metric": "Held-out test examples", "Value": len(y_true)},
    {"Metric": "Number of disease classes", "Value": num_classes},
    {"Metric": "Test loss", "Value": test_loss},
    {"Metric": "Test accuracy", "Value": test_accuracy},
    {"Metric": "Macro precision", "Value": test_precision},
    {"Metric": "Macro recall", "Value": test_recall},
    {"Metric": "Macro F1", "Value": test_f1},
    {"Metric": "Top-3 accuracy", "Value": top3_accuracy}
])

summary.to_csv(OUT / "overall_scores.csv", index=False)

with open(OUT / "submission_notes.txt", "w") as f:
    f.write("Disease Prediction using Fine-Tuned BERT\n")
    f.write("Dataset: gretelai/symptom_to_diagnosis\n")
    f.write("Model: BERT-base-uncased, fine-tuned for 22 classes\n")
    f.write("Training: 3 epochs in Experiment 1; 5 epochs in Experiment 2\n")
    f.write("Experiment 2 test scores are computed from the corrected saved checkpoint.\n")
    f.write("The training loss was not logged per epoch; no training-loss curve is fabricated.\n")
    f.write("This is an educational classifier and not clinically validated.\n\n")
    f.write(comparison.to_string(index=False))
    f.write("\n\n")
    f.write(summary.to_string(index=False))

# ==========================================================
# 9. ZIP ALL OUTPUT FILES AND DOWNLOAD
# ==========================================================
zip_path = "/content/disease_prediction_submission.zip"

with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as z:
    for path in OUT.rglob("*"):
        if path.is_file():
            z.write(path, arcname=path.relative_to(OUT))

print("\nAll outputs saved to:", OUT)
print("ZIP package:", zip_path)

files.download(zip_path)
```
